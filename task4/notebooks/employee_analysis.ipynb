# Employee Data Analysis

## Introduction
This notebook contains a simple analysis of employee data.

## Objectives
- Read employee information
- Calculate department-wise employee count
- Find average salary
- Check employees from Chennai

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Read employee data
df = pd.read_csv(r"C:\Users\HP\Desktop\day6 week assesment\task4\data\employees.csv")

print("Employee Data:")
display(df.head())
```

## Employee Data Analysis

```python
print("Total employees:", len(df))

print("\nDepartment-wise employee count:")
print(df["department"].value_counts())

print("\nAverage salary:")
print(df["salary"].mean())

print("\nEmployees from Chennai:")
print(df[df["city"] == "Chennai"])
```

```python
%timeit df["salary"].mean()
```
